```python
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas
from reportlab.lib.utils import ImageReader
```

```python
# Datos de la institución
nombre_institucion = "UNIVERSIDAD NACIONAL AUTÓNOMA DE MÉXICO "
direccion = "ESCUELA NACIONAL PREPARATORIA"
telefono = "PLANTEL 4 VIDAL CASTAÑEDA Y NÁJERA"

# Archivos de los logos
logo1_path = "/content/download.png"
logo2_path = "/content/ENP4_PNG_azul.png"
```

```python
# Crear el PDF
def crear_pdf(output_path):
    c = canvas.Canvas(output_path, pagesize=letter)
    width, height = letter

    # Tamaño de los logos
    logo1_width = 110  # Aumentar el tamaño del logo 1
    logo1_height = 120
    logo2_width = 70   # Disminuir el tamaño del logo 2
    logo2_height = 70

    # Posiciones de los logos (ajustadas para bajar los logos un poco)
    logo1_x = 50
    logo1_y = height - 150
    logo2_x = width - 130
    logo2_y = height - 130


    # esto ya quedó

    # Dibujar los logos

    logo1 = ImageReader(logo1_path)
    logo2 = ImageReader(logo2_path)
    c.drawImage(logo1, logo1_x, logo1_y, width=logo1_width, height=logo1_height)
    c.drawImage(logo2, logo2_x, logo2_y, width=logo2_width, height=logo2_height)

    # Centrar los datos de la institución entre los logos
    text = c.beginText()
    text.setFont("Helvetica", 11)

    # Calcular la posición centrada entre los logos
    center_x = (logo1_x + logo1_width + logo2_x) / 2


    text.setTextOrigin(center_x - c.stringWidth(nombre_institucion) / 2, height - 100)
    text.textLine(nombre_institucion)

    text.setTextOrigin(center_x - c.stringWidth(direccion) / 2, height - 120)
    text.textLine(direccion)

    text.setFont("Helvetica", 10)
    text.setTextOrigin(center_x - c.stringWidth(telefono) / 2, height - 140)
    text.textLine(telefono)

    c.drawText(text)
    c.save()

# Crear el PDF con el nombre deseado
crear_pdf("encabezado.pdf")
```

```python
import pandas as pd
import xlsxwriter

# Crear un DataFrame vacío con las columnas necesarias
horas = [
    "07:00 - 07:50", "07:50 - 08:40", "08:40 - 09:30", "09:30 - 10:20",
    "10:20 - 11:10", "11:10 - 12:00", "12:00 - 12:50", "12:50 - 13:40",
    "13:40 - 14:30", "14:30 - 15:20", "15:20 - 16:10", "16:10 - 17:00",
    "17:00 - 17:50", "17:50 - 18:40", "18:40 - 19:30", "19:30 - 20:20",
    "20:20 - 21:10", "21:10 - 22:00"
]
dias = ["LUNES", "MARTES", "MIÉRCOLES", "JUEVES", "VIERNES"]

# Crear un DataFrame con las horas como índice y los días como columnas
horario_df = pd.DataFrame(index=horas, columns=dias)

# Función para llenar el horario de una persona (ejemplo)
def llenar_horario(persona, horario_df):
    # Llenar con algunas materias de ejemplo
    horario_df.loc["07:00 - 07:50", "LUNES"] = "Materia A / C101"
    horario_df.loc["10:20 - 11:10", "MIÉRCOLES"] = "Materia B / C102"
    horario_df.loc["16:10 - 17:00", "JUEVES"] = "Materia C / C103"
    # Puedes agregar más asignaciones según sea necesario
    return horario_df

# Crear un archivo Excel con el horario de 250 personas
with pd.ExcelWriter('horarios.xlsx', engine='xlsxwriter') as writer:
    for i in range(1, 251):
        persona = f'Persona {i}'
        df_persona = llenar_horario(persona, horario_df.copy())
        df_persona.to_excel(writer, sheet_name=f'Persona {i}')

print("El archivo 'horarios.xlsx' ha sido creado.")
```
